```python
#find the sum of all even numbers between 1-100
total = 0
for i in range(1,101):
  if i % 2 == 0:
    total = total + i
print(total)
```

```python
# count the even numbers from 1 to 20
count = 0
for i in range(1,21):
  if i % 2 == 0:
    count = count + 1
print(count)
```

```python
#COunt numbers divisible by both 3 and 5 between 1 and 100
count = 0
for i in range(1,101):
  if i % 3 == 0 and i % 5 == 0:
    count += 1 #count = count + 1
print(count)
```

```python
# ask user for number of students, add the marks of each student
student = int(input('Enter number of students: '))
total = 0
for i in range(student):
  marks = float(input('Enter marks: '))
  total = total + marks # total += marks
print('Total:',total)
print('Average:', total/student)
```

```python
# now add pass/fail count to the above code, pass : 50
student = int(input('Enter number of students: '))
total = 0
passed = 0
failed = 0
for i in range(student):
  marks = float(input('Enter marks: '))
  total = total + marks # total += marks
  if marks >= 50:
    passed = passed + 1
  else:
    failed = failed + 1
print('Total:',total)
print('Average:', total/student)
print('Students passed:',passed)
print('Students failed:',failed)
```
