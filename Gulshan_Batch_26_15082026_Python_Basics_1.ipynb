Variable naming rules in Python:

1. variable name can contain any letter, numbers or underscore _
2. name can not start with a number: 1st_name XX
3. name can not contain spaces: student name XX: student_name (snake case), studentName (camel case)
- snake case this_sentence_is_written_in_snake_case
- camel case: thisSentenceIsWrittenInCamelCase
4. name can not be a python reserver keyword
