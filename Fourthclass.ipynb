```python
#Ask user for a number ,print wheter a number @ is even or odd
name = int(input('Enter a number:'))
if name % 2 == 0:
    print('Even')
else:
    print('Odd')
```

```python
#Ask user for a number and display negative or positive 
#use elif
number = int(input('Enter a number:'))
if number > 0:
    print('Positive')
elif number==0:
    print('Whole Number')
else:
    print('Negative')
```

```python
#Ask user for marks and print the grade
#90+ > A
#80-89 > B
#70-79 > C
#60-69 > D
#Below 60 is F
marks = float(input('Enter your marks:'))
if marks >=90:
    print('A')
elif marks >=80:
    print('B')
elif marks >= 70:
    print('C')
elif marks >= 60:
    print('D')
else:
    print('You are Fail,Better Luck Next time')
```

```python
#ask user for temperature and display the wheather
#40 or above > very Hot
#30-39       >Hot
#20-29       >Pleasent
#Below 20    >cold
userTemp= float(input('Enter the temperature:'))
if userTemp >= 40:
    print('Very Hot')
elif userTemp >=30:
    print('Hot')
elif userTemp >=20:
    print('Pleasent')
else:
    print('Cold')
```
