```python
# ask user for a word, count the number of 'a'
word = input('Enter a word: ')
count = 0
for letter in word:
  if letter == 'a':
    count += 1
print('Number of a:',count)
```

```python
# vowels: a,e,i,o,u
# ask user for a word and count vowels
word = input('Enter a word: ').lower()
count = 0
for letter in word:
  if letter =='a' or letter=='e' or letter=='i' or letter =='o' or letter=='u':
    count += 1
print('Number of vowels:',count)
```

```python
# ask user for a word and display no. of chars,
# no. of vowels, no. of consonants, no. of a's
word = input('Enter a word: ')
vowel = consonant = characters = letter_a = 0
for letter in word:
  characters += 1
  if letter in 'aeiouAEIOU':
    vowel += 1
  else:
    consonant += 1
  if letter == 'a':
    letter_a += 1
print('Consonants:',consonant)
print('Vowels:',vowel)
print('Characters:',characters)
print("Number of a's:",letter_a)
```
