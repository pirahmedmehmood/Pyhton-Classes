```python
# Last class problem
# ask user number of items, price of each item, add to list
# calculate total, discount, final price
# if total > 2500, apply 10% discount
times = int(input('How many items do you want? '))
price = []
for i in range(times):
  product = int(input('Enter price: '))
  price.append(product)
total = sum(price)
if total >= 2500:
  discount = total * 10 / 100
else:
  discount = 0
print('Total',total)
print('Discount', discount)
print('Final Price',total - discount)
```
