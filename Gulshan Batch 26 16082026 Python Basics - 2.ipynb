```python
# ask user for name, marks in maths, chem, english
# display name, total marks, avg marks
name = input('Enter your name:')
maths = float(input('Enter mathematics marks'))
chem = float(input('Enter chemistry marks'))
eng = float(input('Enter english marks'))

total = maths + chem + eng
avg = total / 3

print('Welcome',name)
print('Your total marks are',total)
print('Your average marks are',avg)
```
