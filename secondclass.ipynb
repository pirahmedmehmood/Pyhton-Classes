```python
name= input('Enter your name:')
marksin_maths= int(input('Enter your marks in maths:'))
marksin_chem= int(input("Enter your markis in chem:"))
marksin_english=int(input('Enter your marks in eng:'))
total= marksin_maths + marksin_chem + marksin_english
avg = total / 3
print('Name:',name)
print('MAths',marksin_maths)
print('Chem:',marksin_chem)
print('English:',marksin_english)
print('Total:',total)
print('Avg Marks:',avg)
```
