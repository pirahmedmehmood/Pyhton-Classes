```python
# print the total marks of each student

for student in marks:
  total = 0
  for mark in student:
    total += mark
  print(total)
```

```python
# Find the average marks of each student. Use nested for
for student in marks:
  total = 0
  for mark in student:
    total += mark
  print(total/3)
```

```python
students = [
    ["Ali", 80, 75, 88],
    ["Ahmed", 72, 81, 76],
    ["Sara", 91, 89, 95]
]
# find the total marks of each student. Use nested for loop
for student in students:
  total = 0
  for marks in student[1:]:
    total += marks
  print(student[0],total)
```
