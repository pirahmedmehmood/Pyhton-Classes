```python
#ask withdrawl
#ask user:
#enter account balance
#enter withdrawl amount

#rules:
#if withdrawl amount is greater than balance
#insufficient balance
#otherwise calculate remaining balance
#if remaining balance is below 1000:
#warning :low balance 
#otherwise: transaction successfull 
```

```python
balance =float(input('Enter account balance:'))
withdraw=float(input('Enter withdrawl amount:'))
if withdraw>balance:
    print('Insufficient amount')
elif withdraw-balance<1000:
    print('You have withdraw',withdraw)
    print('Low Balance')
else:
    print('Transaction Successfull')
```

```python
#A university student can enroll in a course 
#if they have completed the prerequisite course and their fees is paid
```
